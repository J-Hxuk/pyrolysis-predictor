# 02. 학습·평가 (Train & Evaluate)

이 노트북은 `pyrolysis.training` 과 `pyrolysis.evaluation` 의 공통 함수를 호출한다.  
모델 클래스나 평가 로직을 이 노트북에 복사하지 않는다.  
합성 설정 점검은 CLI `python -m pyrolysis.cli train --demo` 과 **같은 공통 함수**를 호출한다.

**커밋 전: Kernel → Restart & Clear Output 후 저장.** 실제 연구 셀 출력을 저장하지 않는다.

````python
from pathlib import Path

from pyrolysis.config import load_problem
from pyrolysis.training import run_demo_training

# 합성 설정 점검. 실제 학습·평가 계획은 작업 27·28 에서 확정한다.
problem_path = Path('configs/demo.yaml')
output_root = Path('runs')

spec = load_problem(problem_path)
````

````python
# run_demo_training 은 학습·평가 자료를 명시적으로 분리하고
# 전처리를 학습 자료에만 fit 한다 (전처리 누출 방지). CLI 와 같은 함수.
result = run_demo_training(spec, output_root)

print('run_id:', result['run_id'])
print('학습 행:', result['n_train'], '/ 평가 행:', result['n_eval'])
print('MAE :', round(result['mae'], 4))
print('RMSE:', round(result['rmse'], 4))
print('기록 파일:', result['record_path'])
print('***', result['disclaimer'], '***')
````

같은 학습을 CLI 로 실행:

```powershell
python -m pyrolysis.cli train --demo --problem configs/demo.yaml --output runs
```

동일 합성 설정·난수이면 노트북과 CLI 의 MAE·RMSE 가 일치한다.  
실제 대상별 비교·선정·독립 평가는 작업 13·14 의 함수를 작업 28 에서 실제 설정으로 호출한다.
